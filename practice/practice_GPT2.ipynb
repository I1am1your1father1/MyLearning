```python
# Basic gengeration
from transformers import pipeline, AutoTokenizer
```

```python
# Use GPT-2 model for text generation
tokenizer = AutoTokenizer.from_pretrained('gpt2')
tokenizer.pad_token = tokenizer.eos_token

generator = pipeline('text-generation', model='gpt2', tokenizer=tokenizer)
prompt = "人工智能的未来发展"

output = generator(prompt,
                   max_length=100,
                   truncation=True,
                   pad_token_id=tokenizer.eos_token_id)
```

```python
# Chinese is not supported by GPT-2
print(output[0]['generated_text'])
```

```python
# Try English prompt
prompt = "The future development of artificial intelligence"

output = generator(prompt,
                   max_length=100,
                   truncation=True,
                   pad_token_id=tokenizer.eos_token_id)
```

```python
# It has good generation in English
print(output[0]['generated_text'])
```
