```python
import jieba
import hanlp
import spacy
```

```python
text = "自然语言处理是人工智能的重要分支，近年来发展迅速。"
english_text = "Natural Language Processing is amazing."
```

```python
# 1. 使用jieba分词
words = list(jieba.cut(text))
print("分词结果:", words)
```

```python
# 2. 使用HanLP进行词性标注
tok_model = hanlp.load(hanlp.pretrained.tok.COARSE_ELECTRA_SMALL_ZH)
hanlp_seg = tok_model(text)
pos_model = hanlp.load(hanlp.pretrained.pos.CTB9_POS_ELECTRA_SMALL)
hanlp_pos = pos_model(hanlp_seg)

for word, pos in zip(hanlp_seg, hanlp_pos):
    print(f"{word}/{pos}", end=" ")
```

```python
# 3. 使用spaCy的英文模型处理英文部分
nlp = spacy.load("en_core_web_sm")
doc = nlp(english_text)
print("\n英文实体识别:")
for ent in doc.ents:
    print(ent.text, ent.label_)
```
